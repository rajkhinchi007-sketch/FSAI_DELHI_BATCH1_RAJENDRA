```python
import gradio as gr
import datetime
import re

print("Gradio version:", gr.__version__)
print("Libraries imported successfully!")
```

```python
def chatbot_response(message, history):

    message = message.lower().strip()

    if not message:
        return "Please enter a message."

    # Greetings
    if any(word in message for word in ["hello", "hi", "hey"]):
        return "Hello! Welcome to Rajendra's AI Assistant. How can I help you today?"

    # Name
    elif "your name" in message:
        return "I am Rajendra's AI Assistant, a chatbot developed using Python and Gradio."

    # Python
    elif "python" in message:
        return (
            "Python is a popular programming language used in "
            "AI, Machine Learning, Data Science, Web Development "
            "and Automation."
        )

    # Artificial Intelligence
    elif "artificial intelligence" in message or "what is ai" in message:
        return (
            "Artificial Intelligence (AI) is a technology that enables "
            "machines to perform tasks that normally require human intelligence, "
            "such as learning, reasoning and problem-solving."
        )

    # Machine Learning
    elif "machine learning" in message:
        return (
            "Machine Learning is a branch of AI that enables computers "
            "to learn patterns from data and make predictions."
        )

    # Data Science
    elif "data science" in message:
        return (
            "Data Science combines statistics, programming and domain "
            "knowledge to extract useful insights from data."
        )

    # Current date
    elif "date" in message or "time" in message:
        return "Current system date and time: " + datetime.datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    # Thank you
    elif "thank" in message:
        return "You're welcome! I am happy to help."

    # Goodbye
    elif "bye" in message or "goodbye" in message:
        return "Goodbye! Thank you for using Rajendra's AI Assistant."

    # Default response
    else:
        return (
            "That's an interesting question! I am a demonstration chatbot "
            "with predefined responses. Try asking about Python, Artificial "
            "Intelligence, Machine Learning or Data Science."
        )

print("Chatbot response function created successfully!")
```

```python
demo = gr.ChatInterface(
    fn=chatbot_response,

    title="Rajendra's AI Assistant",

    description="""
    Welcome to Rajendra's AI Assistant!

    This interactive chatbot is developed using Python and Gradio.

    You can ask questions about:
    - Python Programming
    - Artificial Intelligence
    - Machine Learning
    - Data Science
    - General Technical Topics

    Type your question below or select one of the example prompts.
    """,

    examples=[
        "Hello! How are you?",
        "What is Artificial Intelligence?",
        "Explain Machine Learning.",
        "What is Python programming?"
    ],

    chatbot=gr.Chatbot(
        height=450,
        placeholder="Welcome! Start your conversation here."
    ),

    textbox=gr.Textbox(
        placeholder="Type your question here...",
        container=True,
        scale=7
    )
)

print("Gradio interface created successfully!")
```

```python
demo.launch(
    share=False,
    inbrowser=True
)
```
