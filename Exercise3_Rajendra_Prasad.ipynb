```python
import os
import cv2
import matplotlib.pyplot as plt
from IPython.display import display, Markdown, Video

print("Exercise 3: AI Video Synthesis")
print("Libraries imported successfully!")
```

```python
# Define Exercise 2 image directory
image_folder = "Exercise2_RAJENDRA_PRASAD_Images"

# Define image paths
image1_path = os.path.join(image_folder, "Original_Image1.jpg")
image2_path = os.path.join(image_folder, "Original_Image2.jpg")
blended_path = os.path.join(image_folder, "Blended_Image.jpg")

# Check image availability
image_paths = [image1_path, image2_path, blended_path]

for path in image_paths:
    if os.path.exists(path):
        print("Available:", path)
    else:
        print("Missing:", path)
```

```python
# Load images using OpenCV
image1 = cv2.imread(image1_path)
image2 = cv2.imread(image2_path)
blended_image = cv2.imread(blended_path)

# Validate images
if any(img is None for img in [image1, image2, blended_image]):
    raise FileNotFoundError(
        "One or more images could not be loaded. "
        "Please verify the Exercise 2 image files."
    )

# Convert BGR to RGB
image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended_image, cv2.COLOR_BGR2RGB)

# Display images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image1_rgb)
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(image2_rgb)
plt.title("Original Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis("off")

plt.suptitle("Input Images for AI Video Generation",
             fontsize=15, fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
# Selected Generative AI video tool

tool_name = "Google Veo"

tool_website = "https://deepmind.google/models/veo/"

video_duration = 3

print("Selected AI Tool:", tool_name)
print("Official Website:", tool_website)
print("Required Video Duration:", video_duration, "seconds")
```

```python
# Structured video generation prompt

video_prompt = """
Create a visually appealing, cinematic 3-second video
using the three provided reference images.

SCENE 1:
Begin by displaying the first original photograph.
Apply a smooth cinematic zoom-in effect with natural lighting.

SCENE 2:
Gradually transition from the first photograph to the
second photograph using a smooth cross-dissolve effect.

SCENE 3:
Transition into the third reference image, which is the
composite image created by blending the two original images.

CAMERA INSTRUCTIONS:
Use smooth camera movements, subtle zoom effects,
stable framing and professional cinematic transitions.

VISUAL QUALITY:
Maintain the original visual characteristics, colors,
objects and details of all three reference images.
Use realistic lighting, balanced contrast and high-quality visuals.

FINAL OUTPUT:
The video must be approximately 3 seconds long.
The final frame should clearly display the blended composite image.
Avoid unnecessary text, subtitles, logos or watermarks.
"""

print("AI Video Generation Prompt:")
print("=" * 70)
print(video_prompt)
```

```python
display(Markdown("# AI Video Generation Prompt"))

display(Markdown(f"**Selected Tool:** {tool_name}"))

display(Markdown(f"**Duration:** {video_duration} seconds"))

display(Markdown("### Exact Prompt"))

display(Markdown(f"> {video_prompt}"))
```

```python
camera_instructions = {
    "Duration": "3 Seconds",
    "Opening Scene": "Original Image 1",
    "Middle Scene": "Original Image 2",
    "Final Scene": "Blended Image",
    "Camera Movement": "Smooth Cinematic Zoom",
    "Transition": "Cross-Dissolve",
    "Visual Style": "Realistic and Cinematic",
    "Lighting": "Natural and Balanced",
    "Final Output": "3-Second AI Generated Video"
}

print("Camera and Transition Instructions")
print("-" * 45)

for key, value in camera_instructions.items():
    print(f"{key}: {value}")
```

```python
# Create output directory
output_folder = "Exercise3_Rajendra_Prasad"
os.makedirs(output_folder, exist_ok=True)

# Save exact prompt
prompt_file = os.path.join(
    output_folder,
    "AI_Video_Generation_Prompt.txt"
)

with open(prompt_file, "w", encoding="utf-8") as file:
    file.write("EXERCISE 3: GENERATIVE AI VIDEO SYNTHESIS\n")
    file.write("=" * 60 + "\n\n")
    file.write("Participant: Rajendra Prasad\n")
    file.write("Selected Tool: Google Veo\n")
    file.write("Video Duration: 3 Seconds\n\n")
    file.write("EXACT VIDEO GENERATION PROMPT\n")
    file.write("-" * 60 + "\n")
    file.write(video_prompt)
    file.write("\n\nCAMERA AND TRANSITION INSTRUCTIONS\n")

    for key, value in camera_instructions.items():
        file.write(f"{key}: {value}\n")

print("Prompt documentation saved successfully!")
print("File:", prompt_file)
```
