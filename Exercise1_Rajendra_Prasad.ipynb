```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Generate random numerical data
data = {
    'f1': np.random.randint(10, 100, 100),
    'f2': np.random.randint(20, 150, 100),
    'f3': np.random.randint(30, 200, 100),
    'f4': np.random.randint(40, 250, 100)
}

# Create DataFrame
df = pd.DataFrame(data)

# Display first five rows
print("Generated DataFrame:")
display(df.head())

# Display last five rows
print("Generated DataFrame:")
display(df.tail())

# Display dataset information
print("\nDataset Information:")
print(df.info())
```

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    df['f2'],
    df['f3'],
    color='blue',
    alpha=0.7,
    edgecolors='black'
)

plt.xlabel('Feature f2', fontsize=12)
plt.ylabel('Feature f3', fontsize=12)

plt.title(
    'Scatter Plot: Relationship Between f2 and f3',
    fontsize=14,
    fontweight='bold'
)

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Save scatter plot
plt.savefig('Exercise1_ScatterPlot.png', dpi=300)

plt.show()
```

```python
# Calculate arithmetic mean across four columns
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display first 5 updated DataFrame
print("Updated DataFrame with f5:")
display(df.head(5))

# Display last 5 updated DataFrame
print("Updated DataFrame with f5:")
display(df.tail(5))

# Display dataset information
print("\nDataset Information:")
print(df.info())
```

```python
#Perform statistical analysis
print("Statistical Summary of Dataset:")
display(df.describe())

print("\nAverage values of all columns:")
display(df.mean())
```

```python
# Export complete DataFrame to Excel
output_file = 'Exercise1_Rajendra_Prasad.xlsx'

df.to_excel(output_file, index=False, engine='openpyxl')

print("Excel file successfully generated!")
print("File Name:", output_file)
print("Total Records:", len(df))
print("Total Columns:", len(df.columns))
```

```python
# Read the exported Excel file
verified_df = pd.read_excel(output_file)

print("Excel File Verification:")
display(verified_df.head())

assert verified_df.shape == (100, 5)

print("Verification successful!")
```
